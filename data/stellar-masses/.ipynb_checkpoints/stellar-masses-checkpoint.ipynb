```python
import pandas as pd

df = pd.read_csv(
    "stellar-masses.txt",
    skiprows=16,
    sep=r"\s+",
    header=None,
    engine="python"
)

df.info()
```

```python
df.columns = ["Galaxy", "logL", 'logM', 'e_M', 'color', 'M_L']
df.head()
```

### Train/Test Data Creation

```python
# Split df into test and train
dftrain = df.sample(int(len(df)//1.25), random_state=42)
dftest = df.drop(dftrain.index)

dftrain.to_csv('stellar-masses-train.csv', index = False)
dftest.to_csv('stellar-masses-test.csv', index = False)
```

```python
import shutil
import os
os.makedirs('../train', exist_ok = True)
os.makedirs('../test', exist_ok = True)

shutil.move('stellar-masses-train.csv', '../train/stellar-masses.csv')
shutil.move('stellar-masses-test.csv', '../test/stellar-masses.csv')
```
