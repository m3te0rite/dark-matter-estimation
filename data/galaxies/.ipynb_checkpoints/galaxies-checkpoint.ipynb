```python
import pandas as pd

df = pd.read_csv(
    "galaxies.txt",
    skiprows=98,
    sep=r"\s+",
    header=None,
    engine="python"
)

df.info()
```

```python
df.columns = [
    "Galaxy", "T", "D", "e_D", "f_D", "Inc", "e_Inc", "L_3_6", "e_L_3_6", "Reff", "SBeff", "Rdisk", "SBdisk", "MHI", "RHI", "Vflat", "e_Vflat", "Q", "Ref"]
df.head()
```

### Train/Test Data Creation

```python
# Split df into test and train
dftrain = df.sample(int(len(df)//1.25), random_state=42)
dftest = df.drop(dftrain.index)

dftrain.to_csv('galaxies-train.csv', index = False)
dftest.to_csv('galaxies-test.csv', index = False)
```

```python
import shutil
import os
os.makedirs('../train', exist_ok = True)
os.makedirs('../test', exist_ok = True)

shutil.move('galaxies-train.csv', '../train/galaxies.csv')
shutil.move('galaxies-test.csv', '../test/galaxies.csv')
```
