```python
import pandas as pd

df = pd.read_csv(
    "newtonian-mass-models.txt",
    skiprows=25,
    sep=r"\s+",
    header=None,
    engine="python"
)

df.info()
```

```python
df.columns = ["ID", "D", 'R', 'Vobs', 'e_Vobs', 'Vgas', 'Vdisk', 'Vbul', 'SBdisk', 'SBbul']
df.head()
```

### Train/Test Data Creation

```python
# Split df into test and train
dftrain = df.sample(int(len(df)//1.25), random_state=42)
dftest = df.drop(dftrain.index)

dftrain.to_csv('mass-train.csv', index = False)
dftest.to_csv('mass-test.csv', index = False)
```

```python
import shutil
import os
os.makedirs('../train', exist_ok = True)
os.makedirs('../test', exist_ok = True)

shutil.move('mass-train.csv', '../train/newt-mass.csv')
shutil.move('mass-test.csv', '../test/newt-mass.csv')
```
