```python
import pandas as pd

df = pd.read_csv(
    "Baryonic-Tully-Fisher Relation.txt",
    skiprows=30,
    sep=r"\s+",
    header=None,
    engine="python"
)

df.info()
```

```python
df.columns = ["Name", "log_Mb", "e_log_Mb", "Inc", "e_Inc", "Vf", "e_Vf", "V2exp", "e_V2exp", "V2eff", "e_V2eff", "Vmax", "e_Vmax", "Wp20", "e_Wp20", "Wm50", "e_Wm50", "Wm50c", "e_Wm50c"]
df.head()
```

### Train/Test Data Creation

```python
# Split df into test and train
dftrain = df.sample(int(len(df)//1.25), random_state=42)
dftest = df.drop(dftrain.index)

dftrain.to_csv('BTF-train.csv', index = False)
dftest.to_csv('BTF-test.csv', index = False)
```

```python
import shutil
import os
os.makedirs('../train', exist_ok = True)
os.makedirs('../test', exist_ok = True)

shutil.move('BTF-train.csv', '../train/BTF.csv')
shutil.move('BTF-test.csv', '../test/BTF.csv')
```
